```python
#20 image cartoonnification
import cv2
import matplotlib.pyplot as plt
# Read image
path = r"C:\DIPIMG\img2.jpg"
img = cv2.imread(path)
# Convert BGR to RGB
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
# Resize image
img = cv2.resize(img, (500,500))

# Apply cartoon/stylization effect
cartoon = cv2.stylization(
    img,
    sigma_s=200,
    sigma_r=0.25
)
# Display images
plt.figure(figsize=(12,6))
# Original Image
plt.subplot(1,2,1)
plt.imshow(img)
plt.title("Original Image")
plt.axis('off')
# Cartoon Image
plt.subplot(1,2,2)
plt.imshow(cartoon)
plt.title("Cartoon Effect")
plt.axis('off')
plt.show()
```

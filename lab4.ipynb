```python
#pixel-wise operations using Euclidean distance

import cv2
import math
import matplotlib.pyplot as plt

# 1. Read grayscale image
# Replace this path with the actual location of your image
path = r"C:\DIPIMG\img5.jpg"
img = cv2.imread(path, 0)

# 2. Select two pixels (row, column coordinates)
x1, y1 = 190, 190
x2, y2 = 350, 350

# 3. Get pixel intensity values
pixel1 = img[x1, y1]
pixel2 = img[x2, y2]

# 4. Display pixel information
print("\nFirst Pixel Position  : (", x1, ",", y1, ")")
print("Second Pixel Position : (", x2, ",", y2, ")")
print("\n--- Pixel Intensity Values ---")
print("Pixel 1 value : ", pixel1)
print("Pixel 2 value : ", pixel2)

# 5. Pixel-wise operations (casting to int prevents overflow issues)
addition = int(pixel1) + int(pixel2)
subtraction = abs(int(pixel1) - int(pixel2))

print("\n--- Pixel-wise Operations ---")
print("Addition of pixels    : ", addition)
print("Subtraction of pixels : ", subtraction)

# 6. Distance calculations between the coordinates
# Manhattan Distance: |x1 - x2| + |y1 - y2|
manhattan_dist = abs(x1 - x2) + abs(y1 - y2)

# Euclidean Distance: sqrt((x1 - x2)^2 + (y1 - y2)^2)
euclidean_dist = math.sqrt((x1 - x2)**2 + (y1 - y2)**2)

print("\n--- Distance Calculations ---")
print("Manhattan Distance : ", manhattan_dist)
print("Euclidean Distance : ", euclidean_dist)

# 7. Mark selected pixels on a color version of the image
# Convert grayscale to BGR color space so we can draw colored markers
img_color = cv2.cvtColor(img, cv2.COLOR_GRAY2BGR)

# OpenCV uses (x, y) as (column, row), so we swap the indices to (y, x)
cv2.circle(img_color, (y1, x1), 10, (255, 0, 0), -1)
cv2.circle(img_color, (y2, x2), 10, (255, 0, 0), -1)

# 8. Display the marked image using Matplotlib
plt.figure(figsize=(6, 6))
# Convert BGR to RGB for correct display in Matplotlib
plt.imshow(cv2.cvtColor(img_color, cv2.COLOR_BGR2RGB))
plt.title("Selected Pixels")
plt.axis('off')
plt.show()
```

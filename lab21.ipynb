```python
#21.pencil sketch using edge detection
import cv2
import matplotlib.pyplot as plt
# Read image
path = r"C:\DIPIMG\img20.jpg"
img = cv2.imread(path)
# Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
# Blur image slightly
blur = cv2.GaussianBlur(gray, (3,3), 0)
# Edge detection
edges = cv2.Canny(blur, 50, 180)
# Thicker edges
edges = cv2.dilate(edges, None, iterations=1)
# Invert image
sketch = 255 - edges
# Apply binary threshold for clean black/white sketch
_, sketch = cv2.threshold(sketch, 200, 255, cv2.THRESH_BINARY)
# Display output
plt.figure(figsize=(12,6))
# Original Image
plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis('off')
# Easy Drawing Sketch
plt.subplot(1,2,2)
plt.imshow(sketch, cmap='gray')
plt.title("Easy Sketch")
plt.axis('off')
plt.show()
```

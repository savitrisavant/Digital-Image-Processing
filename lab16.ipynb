```python
# 16. Histogram and Thresholding Operations
# Aim:
# Display histogram and apply different thresholding techniques.

import cv2
import matplotlib.pyplot as plt

# Read grayscale image
path = r"C:\DIPIMG\img2.jpg"

img = cv2.imread(path, 0)

# Thresholding techniques
_, binary = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)
_, binary_inv = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY_INV)
_, trunc = cv2.threshold(img, 127, 255, cv2.THRESH_TRUNC)
_, tozero = cv2.threshold(img, 127, 255, cv2.THRESH_TOZERO)

# Display outputs
plt.figure(figsize=(14, 10))

# 1. Original Image
plt.subplot(2, 3, 1)
plt.imshow(img, cmap='gray')
plt.title("Original")
plt.axis('off')

# 2. Histogram
plt.subplot(2, 3, 2)
plt.hist(img.ravel(), bins=256, range=[0, 256])
plt.title("Histogram")

# 3. Binary
plt.subplot(2, 3, 3)
plt.imshow(binary, cmap='gray')
plt.title("Binary")
plt.axis('off')

# 4. Binary Inverse
plt.subplot(2, 3, 4)
plt.imshow(binary_inv, cmap='gray')
plt.title("Binary Inverse")
plt.axis('off')

# 5. Truncation
plt.subplot(2, 3, 5)
plt.imshow(trunc, cmap='gray')
plt.title("Truncation")
plt.axis('off')

# 6. To Zero
plt.subplot(2, 3, 6)
plt.imshow(tozero, cmap='gray')
plt.title("To Zero")
plt.axis('off')

# Display all images
plt.show()
```

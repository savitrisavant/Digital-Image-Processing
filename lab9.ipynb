```python
#Mean Filter using a 3×3 Averaging Mask  Spatial filtering

import cv2
import matplotlib.pyplot as plt
import numpy as np

path = r"C:\DIPIMG\img7.jpg"
image = cv2.imread(path, 0)
image = cv2.resize(image, (400, 400))

mask = np.array([[1,1,1],[1,1,1],[1,1,1]], dtype=np.float32)
mask = mask / 9
print(mask)

filtered = cv2.filter2D(image, -1, mask)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray')
plt.title("original image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(filtered, cmap='gray')
plt.title("filtered image")
plt.axis("off")

plt.show()
```

```python
#Averaging vs Gaussian Blur (7×7 and 15×15)

import cv2
import matplotlib.pyplot as plt
import numpy as np

path = r"C:\DIPIMG\img7.jpg"
image = cv2.imread(path)
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
image = cv2.resize(image, (400, 400))

avg7 = cv2.blur(image, (7, 7))
acg15 = cv2.blur(image, (15, 15))

gauss7 = cv2.GaussianBlur(image, (7, 7), 0)
gauss15 = cv2.GaussianBlur(image, (15, 15), 0)

plt.figure(figsize=(12, 8))

plt.subplot(2, 3, 1)
plt.imshow(image)
plt.title("original image")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(avg7)
plt.title("average 7X7 image")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(avg7)
plt.title("average 15X15 image")
plt.axis("off")

plt.subplot(2, 3, 4)
plt.imshow(gauss7)
plt.title("guassian 7X7 image")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(gauss15)
plt.title("guassian 15X15 image")
plt.axis("off")

plt.show()


```

```python
#Erosion and Dilation


import cv2
import matplotlib.pyplot as plt
import numpy as np

path = r"C:\DIPIMG\img7.jpg"
image = cv2.imread(path, 0)
_, binary = cv2.threshold(image, 127, 255, cv2.THRESH_BINARY)
kernal = np.ones((5, 5), np.uint8)
erosion = cv2.erode(binary, kernal, iterations=1)
dilation = cv2.dilate(binary, kernal, iterations=1)

plt.figure(figsize=(12, 5))
plt.subplot(1, 3, 1)
plt.imshow(binary, cmap='gray')
plt.title("original image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(erosion, cmap='gray')
plt.title("erosion image")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(dilation, cmap='gray')
plt.title("dilation image")
plt.axis("off")

plt.show()
```

```python
#Morphological Opening and Closing

import cv2
import matplotlib.pyplot as plt
import numpy as np

path = r"C:\DIPIMG\img7.jpg"
image = cv2.imread(path, 0)
_, binary = cv2.threshold(image, 127, 255, cv2.THRESH_BINARY)
kernal = np.ones((5, 5), np.uint8)
opening = cv2.morphologyEx(binary, cv2.MORPH_OPEN, kernal)
closing = cv2.morphologyEx(binary, cv2.MORPH_CLOSE, kernal)

plt.figure(figsize=(10, 8))
plt.subplot(2, 2, 1)
plt.imshow(binary, cmap='gray')
plt.title("original image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(opening, cmap='gray')
plt.title("erosion image")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(closing, cmap='gray')
plt.title("dilation image")
plt.axis("off")

plt.show()
```

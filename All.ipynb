```python
# 1.Geometric Tranformation

import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. Read the image and convert color space from BGR to RGB
path = r"C:\DIPIMG\img14.jpg"  # Replace with your actual image path
img = cv2.imread(path)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Get image dimensions (height and width)
rows, cols = img.shape[:2]

# ==========================================
# 2. Geometric Transformations
# ==========================================

# # Translation (Shifting the image horizontally and vertically by 50 pixels)
M1 = np.float32([[1.5, 0, 50], [0, 1, 50]])
translated = cv2.warpAffine(img, M1, (cols, rows))

# # Rotation (Rotating 45 degrees around the center with a scale factor of 1)
M2 = cv2.getRotationMatrix2D((cols / 2, rows / 2), 45, 1)
rotated = cv2.warpAffine(img, M2, (cols, rows))

# # Scaling (Non-uniform scaling: 80% width, 50% height)
scaled = cv2.resize(img, None, fx=0.8, fy=0.5)

# ==========================================
# 3. Displaying Results
# ==========================================
plt.figure(figsize=(12, 8))

# Original Image
plt.subplot(2, 2, 1)
plt.imshow(img)
plt.title("Original Img")
plt.axis('off')

# Translated Image
plt.subplot(2, 2, 2)
plt.imshow(translated)
plt.title("Translated Img")
plt.axis('off')

# Rotated Image
plt.subplot(2, 2, 3)
plt.imshow(rotated)
plt.title("Rotated Img")
plt.axis('off')

# Scaled Image
plt.subplot(2, 2, 4)
plt.imshow(scaled)
plt.title("Scaled Image")
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
#4.pixel-wise operations using Euclidean distance

import cv2
import math
import matplotlib.pyplot as plt

# 1. Read grayscale image
# Replace this path with the actual location of your image
path = r"C:\DIPIMG\img5.jpg"
img = cv2.imread(path, 0)

# 2. Select two pixels (row, column coordinates)
x1, y1 = 190, 190
x2, y2 = 350, 350

# 3. Get pixel intensity values
pixel1 = img[x1, y1]
pixel2 = img[x2, y2]

# 4. Display pixel information
print("\nFirst Pixel Position  : (", x1, ",", y1, ")")
print("Second Pixel Position : (", x2, ",", y2, ")")
print("\n--- Pixel Intensity Values ---")
print("Pixel 1 value : ", pixel1)
print("Pixel 2 value : ", pixel2)

# 5. Pixel-wise operations (casting to int prevents overflow issues)
addition = int(pixel1) + int(pixel2)
subtraction = abs(int(pixel1) - int(pixel2))

print("\n--- Pixel-wise Operations ---")
print("Addition of pixels    : ", addition)
print("Subtraction of pixels : ", subtraction)

# 6. Distance calculations between the coordinates
# Manhattan Distance: |x1 - x2| + |y1 - y2|
manhattan_dist = abs(x1 - x2) + abs(y1 - y2)

# Euclidean Distance: sqrt((x1 - x2)^2 + (y1 - y2)^2)
euclidean_dist = math.sqrt((x1 - x2)**2 + (y1 - y2)**2)

print("\n--- Distance Calculations ---")
print("Manhattan Distance : ", manhattan_dist)
print("Euclidean Distance : ", euclidean_dist)

# 7. Mark selected pixels on a color version of the image
# Convert grayscale to BGR color space so we can draw colored markers
img_color = cv2.cvtColor(img, cv2.COLOR_GRAY2BGR)

# OpenCV uses (x, y) as (column, row), so we swap the indices to (y, x)
cv2.circle(img_color, (y1, x1), 10, (255, 0, 0), -1)
cv2.circle(img_color, (y2, x2), 10, (255, 0, 0), -1)

# 8. Display the marked image using Matplotlib
plt.figure(figsize=(6, 6))
# Convert BGR to RGB for correct display in Matplotlib
plt.imshow(cv2.cvtColor(img_color, cv2.COLOR_BGR2RGB))
plt.title("Selected Pixels")
plt.axis('off')
plt.show()
```

```python
# 5. Bit Plane Slicing
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read grayscale image
path = r"C:\DIPIMG\img2.jpg"

img = cv2.imread(path, 0)

# Resize image
img = cv2.resize(img, (256, 256))

# Create figure
plt.figure(figsize=(12, 8))

# Display original image
plt.subplot(3, 3, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis('off')

# Extract and display bit planes
for i in range(8):

    # Extract bit plane
    bit_plane = (img >> i) & 1

    # Convert binary values to 0 and 255
    bit_plane = bit_plane * 255

    # Display bit plane
    plt.subplot(3, 3, i + 2)
    plt.imshow(bit_plane, cmap='gray')

    plt.title(f"Bit Plane {i}")
    plt.axis('off')

# Show all images
plt.tight_layout()
plt.show()
```

```python
# 7. Image Quantization
# Reduce gray levels in an image.

import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read image
path = r"C:\DIPIMG\img3.jpg"
img = cv2.imread(path, 0)

# Quantization function
def quantize(image, levels):
    factor = 256 // levels
    return (image // factor) * factor

# Different gray levels
img128 = quantize(img, 128)
img32 = quantize(img, 32)
img8 = quantize(img, 8)
img2 = quantize(img, 2)

# Display
images = [img, img128, img32, img8, img2]
titles = ["Original", "128 Levels", "32 Levels", "8 Levels", "2 Levels"]
plt.figure(figsize=(15,8))
for i in range(5):
    plt.subplot(2,3,i+1)
    plt.imshow(images[i], cmap='gray')
    plt.title(titles[i])
    plt.axis('off')
plt.show()
```

```python
#19 edge linking

import cv2
import matplotlib.pyplot as plt
# Read image
path = r"C:\DIPIMG\img13.jpg"
img = cv2.imread(path)
# Convert to RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
# Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
# Canny Edge Detection
edges = cv2.Canny(gray, 100, 200)
# Find contours
contours, _ = cv2.findContours(edges,
                               cv2.RETR_EXTERNAL,
                               cv2.CHAIN_APPROX_SIMPLE)
# Draw boundaries
boundary = img_rgb.copy()
cv2.drawContours(boundary, contours, -1, (255,0,0), 2)
# Display outputs
plt.figure(figsize=(15,5))
# Original Image
plt.subplot(1,3,1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis('off')
# Edge Detection
plt.subplot(1,3,2)
plt.imshow(edges, cmap='gray')
plt.title("Edge Detection")
plt.axis('off')
# Boundary Detection
plt.subplot(1,3,3)
plt.imshow(boundary)
plt.title("Boundary Detection")
plt.axis('off')
plt.show()
```

```python
#21.pencil sketch using edge detection
import cv2
import matplotlib.pyplot as plt
# Read image
path = r"C:\DIPIMG\img20.jpg"
img = cv2.imread(path)
# Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
# Blur image slightly
blur = cv2.GaussianBlur(gray, (3,3), 0)
# Edge detection
edges = cv2.Canny(blur, 50, 180)
# Thicker edges
edges = cv2.dilate(edges, None, iterations=1)
# Invert image
sketch = 255 - edges
# Apply binary threshold for clean black/white sketch
_, sketch = cv2.threshold(sketch, 200, 255, cv2.THRESH_BINARY)
# Display output
plt.figure(figsize=(12,6))
# Original Image
plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis('off')
# Easy Drawing Sketch
plt.subplot(1,2,2)
plt.imshow(sketch, cmap='gray')
plt.title("Easy Sketch")
plt.axis('off')
plt.show()
```

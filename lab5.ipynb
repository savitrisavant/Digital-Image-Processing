```python
# 5. Bit Plane Slicing
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read grayscale image
path = r"C:\DIPIMG\img2.jpg"

img = cv2.imread(path, 0)

# Resize image
img = cv2.resize(img, (256, 256))

# Create figure
plt.figure(figsize=(12, 8))

# Display original image
plt.subplot(3, 3, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis('off')

# Extract and display bit planes
for i in range(8):

    # Extract bit plane
    bit_plane = (img >> i) & 1

    # Convert binary values to 0 and 255
    bit_plane = bit_plane * 255

    # Display bit plane
    plt.subplot(3, 3, i + 2)
    plt.imshow(bit_plane, cmap='gray')

    plt.title(f"Bit Plane {i}")
    plt.axis('off')

# Show all images
plt.tight_layout()
plt.show()
```

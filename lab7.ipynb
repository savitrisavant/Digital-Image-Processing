```python
# 7. Image Quantization
# Reduce gray levels in an image.

import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read image
path = r"C:\DIPIMG\img7.jpg"
img = cv2.imread(path, 0)

# Quantization function
def quantize(image, levels):
    factor = 256 // levels
    return (image // factor) * factor

# Different gray levels
img128 = quantize(img, 128)
img32 = quantize(img, 32)
img8 = quantize(img, 8)
img2 = quantize(img, 2)

# Display
images = [img, img128, img32, img8, img2]
titles = ["Original", "128 Levels", "32 Levels", "8 Levels", "2 Levels"]
plt.figure(figsize=(15,8))
for i in range(5):
    plt.subplot(2,3,i+1)
    plt.imshow(images[i], cmap='gray')
    plt.title(titles[i])
    plt.axis('off')
plt.show()
```

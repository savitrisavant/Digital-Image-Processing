```python
#14.Edge Detection

import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read grayscale image
path = r"C:\DIPIMG\img5.jpg"

img = cv2.imread(path, 0)
img = cv2.resize(img, (400, 400))

# Sobel Edge Detection
sobelx = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3)

# Prewitt Edge Detection
kernelx = np.array([[1, 1, 1],
                    [-1, -1, -1]])
prewitt = cv2.filter2D(img, -1, kernelx)

# Roberts Edge Detection
kernel_roberts = np.array([[1, 0], 
                           [0, -1]])
roberts = cv2.filter2D(img, -1, kernel_roberts)

# Canny Edge Detection
canny = cv2.Canny(img, 100, 200)

# Display outputs
plt.figure(figsize=(12, 8))

# 1. Original Image
plt.subplot(2, 3, 1)
plt.imshow(img, cmap='gray')
plt.title("Original")
plt.axis('off')

# 2. Sobel Edge Detection
plt.subplot(2, 3, 2)
plt.imshow(sobelx, cmap='gray')
plt.title("Sobel")
plt.axis('off')

# 3. Prewitt Edge Detection
plt.subplot(2, 3, 3)
plt.imshow(prewitt, cmap='gray')
plt.title("Prewitt")
plt.axis('off')

# 4. Roberts Edge Detection (Completed)
plt.subplot(2, 3, 4)
plt.imshow(roberts, cmap='gray')
plt.title("Roberts")
plt.axis('off')

# 5. Canny Edge Detection (Completed)
plt.subplot(2, 3, 5)
plt.imshow(canny, cmap='gray')
plt.title("Canny")
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
#Geometric Tranformation

import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. Read the image and convert color space from BGR to RGB
path = r"C:\DIPIMG\img5.jpg"  # Replace with your actual image path
img = cv2.imread(path)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Get image dimensions (height and width)
rows, cols = img.shape[:2]

# ==========================================
# 2. Geometric Transformations
# ==========================================

# # Translation (Shifting the image horizontally and vertically by 50 pixels)
M1 = np.float32([[1, 0, 50], [0, 1, 50]])
translated = cv2.warpAffine(img, M1, (cols, rows))

# # Rotation (Rotating 45 degrees around the center with a scale factor of 1)
M2 = cv2.getRotationMatrix2D((cols / 2, rows / 2), 45, 1)
rotated = cv2.warpAffine(img, M2, (cols, rows))

# # Scaling (Non-uniform scaling: 80% width, 50% height)
scaled = cv2.resize(img, None, fx=0.8, fy=0.5)

# ==========================================
# 3. Displaying Results
# ==========================================
plt.figure(figsize=(12, 8))

# Original Image
plt.subplot(2, 2, 1)
plt.imshow(img)
plt.title("Original Img")
plt.axis('off')

# Translated Image
plt.subplot(2, 2, 2)
plt.imshow(translated)
plt.title("Translated Img")
plt.axis('off')

# Rotated Image
plt.subplot(2, 2, 3)
plt.imshow(rotated)
plt.title("Rotated Img")
plt.axis('off')

# Scaled Image
plt.subplot(2, 2, 4)
plt.imshow(scaled)
plt.title("Scaled Image")
plt.axis('off')

plt.tight_layout()
plt.show()
```

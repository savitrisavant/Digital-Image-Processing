```python
#Averaging vs Gaussian Blur (7×7 and 15×15)

import cv2
import matplotlib.pyplot as plt
import numpy as np

path = r"C:\DIPIMG\img7.jpg"
image = cv2.imread(path)
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
image = cv2.resize(image, (400, 400))

avg7 = cv2.blur(image, (7, 7))
acg15 = cv2.blur(image, (15, 15))

gauss7 = cv2.GaussianBlur(image, (7, 7), 0)
gauss15 = cv2.GaussianBlur(image, (15, 15), 0)

plt.figure(figsize=(12, 8))

plt.subplot(2, 3, 1)
plt.imshow(image)
plt.title("original image")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(avg7)
plt.title("average 7X7 image")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(avg7)
plt.title("average 15X15 image")
plt.axis("off")

plt.subplot(2, 3, 4)
plt.imshow(gauss7)
plt.title("guassian 7X7 image")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(gauss15)
plt.title("guassian 15X15 image")
plt.axis("off")

plt.show()


```

```python
#image smoothing and filtering


import cv2
import matplotlib.pyplot as plt

# Provide the actual path to your image file here
path = r"C:\DIPIMG\img6.jpg"

# Read the image and convert from BGR to RGB color space
img = cv2.imread(path)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Apply different smoothing/blurring filters
median = cv2.medianBlur(img, 25)
gaussian = cv2.GaussianBlur(img, (41, 41), 0)
bilateral = cv2.bilateralFilter(img, 19, 200, 200)

# Display the results in a 2x2 grid
plt.figure(figsize=(12, 8))

# 1. Original Image
plt.subplot(2, 2, 1)
plt.imshow(img)
plt.title("Original")
plt.axis('off')

# 2. Median Filtered Image
plt.subplot(2, 2, 2)
plt.imshow(median)
plt.title("Median Filter")
plt.axis('off')

# 3. Gaussian Filtered Image
plt.subplot(2, 2, 3)
plt.imshow(gaussian)
plt.title("Gaussian Filter")
plt.axis('off')

# 4. Bilateral Filtered Image
plt.subplot(2, 2, 4)
plt.imshow(bilateral)
plt.title("Bilateral Filter")
plt.axis('off')

# Render the plot
plt.show()
```

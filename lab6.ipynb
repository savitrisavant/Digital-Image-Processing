```python
# 6. Image Sampling
import cv2
import matplotlib.pyplot as plt

# Read image
path = r"C:\DIPIMG\OIP.webp"
img = cv2.imread(path)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Resize to 256x256
original = cv2.resize(img, (256, 256))

# Sampling
img128 = cv2.resize(original, (128, 128))
img64 = cv2.resize(original, (64, 64))
img32 = cv2.resize(original, (32, 32))

# Resize back
back128 = cv2.resize(img128, (256, 256))
back64 = cv2.resize(img64, (256, 256))
back32 = cv2.resize(img32, (256, 256))

# Display
images = [original, back128, back64, back32]
titles = ["Original", "128x128", "64x64", "32x32"]
plt.figure(figsize=(12, 8))
for i in range(4):
    plt.subplot(2, 2, i + 1)
    plt.imshow(images[i])
    plt.title(titles[i])
    plt.axis('off')
plt.show()
```
